```python
from pathlib import Path
import pandas as pd
```

```python
def load_res_csv():
    repo_root: Path = Path.cwd().parent # Fetch repo root directory
    csv_path: Path = repo_root / "data" / "reservoirs.csv" # Construct path to reservoirs.csv
    
    if not csv_path.exists():
        raise FileNotFoundError(f"CSV file not found at: {csv_path}")
    
    return pd.read_csv(csv_path)

res_df = load_res_csv()
```

```python
data = res_df.copy()[["dato_Id", "fyllingsgrad", "fyllingsgrad_forrige_uke", "endring_fyllingsgrad", "kapasitet_TWh", "fylling_TWh"]] # Dont modify the original dataframe
data["dato_Id"] = pd.to_datetime(data["dato_Id"]) # Convert the "dato_Id" column to datetime format
first_year = data[data["dato_Id"].dt.year == data["dato_Id"].dt.year.min()] # Get the first year in the dataset

print(first_year["dato_Id"].dt.year.unique()) # Get the unique years in the first year dataframe
```

```python
first_month = first_year[first_year["dato_Id"].dt.month ==first_year["dato_Id"].dt.month.min()]
print(first_month["dato_Id"].dt.month.unique()) # Get the unique months in the first month dataframe
```

```python
value_columns = first_month.select_dtypes(include="number").columns # Omit columns with non-numeric data

filtered_df = pd.DataFrame(
    {
        "col": value_columns,
        "first_month": [
            first_month[column].tolist() for column in value_columns
        ],
    }
)
```
